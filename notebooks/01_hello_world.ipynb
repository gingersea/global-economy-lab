# 01 – Hello World: 验证数据获取基本流程

This notebook demonstrates how to use the `global-economy-lab` framework to:
1. Fetch historical price data for S&P 500, CSI 300, and Gold
2. Plot cumulative returns on a single chart
3. Verify that the project environment is working correctly

---
> **Before you start:** Make sure you have run `pip install -r requirements.txt` and (optionally) set `FRED_API_KEY` in your `.env` file.

```python
import sys
from pathlib import Path

# Add project root to path so modules can be imported from any working directory
project_root = Path().resolve().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

print(f"Project root: {project_root}")
```

```python
# Check configuration
from config.settings import get_settings

settings = get_settings()
print(f"Data directory  : {settings.data_dir}")
print(f"Processed cache : {settings.processed_data_dir}")
print(f"FRED API key set: {'Yes' if settings.fred_api_key else 'No – set FRED_API_KEY in .env'}")
```

## Step 1 – Fetch Price Data

```python
from src.data_fetcher.equities import get_index_data
from src.data_fetcher.commodities import get_gold_price

START = "2020-01-01"

print("Fetching S&P 500 …")
spx = get_index_data("^GSPC", start_date=START)
print(f"  {len(spx)} rows fetched")

print("Fetching CSI 300 …")
csi = get_index_data("000300.SH", start_date=START)
print(f"  {len(csi)} rows fetched")

print("Fetching Gold …")
gold = get_gold_price(start_date=START)
print(f"  {len(gold)} rows fetched")
```

## Step 2 – Quick Preview

```python
print("S&P 500 – last 5 rows:")
display(spx.tail())
```

## Step 3 – Plot Cumulative Returns

```python
from src.visualization.dashboard_charts import plot_asset_returns

price_data = {}
if not spx.empty:  price_data["S&P 500"] = spx
if not csi.empty:  price_data["CSI 300"] = csi
if not gold.empty: price_data["Gold"]    = gold

if price_data:
    fig = plot_asset_returns(
        price_data,
        start_date=START,
        backend="plotly",
        title="Cumulative Returns since 2020",
    )
    fig.show()
else:
    print("No data available to plot. Check network connectivity and API keys.")
```
